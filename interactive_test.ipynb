```python
import geopandas as gpd
import shapely
from shapely.geometry import box
```

```python
gdf = gpd.read_file("example_project.json")
```

```python
minx, miny, maxx, maxy = gdf.total_bounds
bounding_polygon = shapely.geometry.box(minx, miny, maxx, maxy)

print(minx, miny, maxx, maxy)

bounding_gdf = gpd.GeoDataFrame(
    geometry=[box(*gdf.total_bounds)],
    crs=gdf.crs
)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 10))

gdf.plot(ax=ax, color="steelblue", edgecolor="black")

bounding_gdf.plot(
    ax=ax,
    facecolor="red",
    edgecolor="darkred",
    alpha=0.2,
    linewidth=2
)

plt.show()
```
