```python
import rasterio
import matplotlib.pyplot as plt
from rasterio.plot import plotting_extent

raster_src = rasterio.open('../imports/elevation_data.tif')

extent = plotting_extent(raster_src.read(1), raster_src.transform)
fig, ax = plt.subplots(figsize=(12, 10))
image = ax.imshow(
    raster_src.read(1),
    extent=extent,
    cmap="terrain",
    origin="upper"
)

cbar = fig.colorbar(image, ax=ax)
cbar.set_label("Elevation (meters)")

plt.show()
```
